```python
list_1 = [1, 2, 3, 4, 5]

list_1.append(6)  # Add an element to the end of the list
print(list_1)
list_1.insert(2, 100)  # Insert an element at a specific index
print(list_1)
```

```python
#delete an element from the list
list_1.remove(100)  # Remove the first occurrence of the specified value
print(list_1)
```

```python
#pop
list_1.pop()  # Remove and return the last element of the list
print(list_1)
```

```python
#reverse
list_1.reverse()  # Reverse the order of elements in the list
print(list_1)
```

```python
list_under_list = [[10, 22, 35], [42, 57, 96], [37, 58, 89]]
list_under_list.sort()
print(list_under_list)
```

```python
#stacking list

list_2 = [3,7,2,5,7,2,5]
list_2.append(12)
list_2.append(23)
list_2.append(15)
list_2.append(17)
list_2.sort()

print(list_2)
```

```python
list_queue = []

list_queue.append(12)
list_queue.append(23)
list_queue.append(15)
list_queue.append(17)

print(list_queue)
```

```python
# list comprehension

even_list = []
for i in range(100): 
        if i%2 == 0:
            even_list.append(i)
print(even_list)
```

```python
random_list = [i for i in range(51) if i%2 != 0]
print(random_list)
```
