```python
inp_fahrenheit = float(input("Enter the Fahrenheit:"))
def fahrenheit_to_celcius(fahrenheit): 
    return (fahrenheit - 32) * 5 / 9

print(f"The celcius is {fahrenheit_to_celcius(inp_fahrenheit)}")

```
