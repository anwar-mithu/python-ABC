```python
#apply function in pandas
#min amx scaling

df = pd.read_csv('students.csv')

mn = df['Total Marks'].min()
mx = df['Total Marks'].max()

scaled = (df['Total Marks'] - mn) / (mx - mn)
# what is scaled? scaled is a new column that contains the min-max scaled values of the Total Marks column. 
# Min-max scaling is a technique used to normalize data by transforming it into a specific range, usually between 0 and 1. 
# This is done by subtracting the minimum value of the column from each value and then dividing by the range (maximum - minimum).
# The resulting values will be in the range [0, 1], 
# which can be useful for machine learning algorithms that are sensitive to the scale of input features.
# a value of scaled is 0.35. it means that the original value of Total Marks is 35% of the way between the minimum and maximum values of the Total Marks column. 
# scaled, df['Total Marks'][0]

df ['Scaled Marks'] = scaled
df.head(10)
df.to_csv('students.csv', index=False)
```

```python
# maxc than 200 marks are A+ grade

def grading_system(marks): 
    if marks > 200:
        return 'A+'
    elif marks > 180:
        return 'A'
    elif marks > 160:
        return 'B+'
    elif marks > 140:
        return 'B'
    elif marks > 90:
        return 'C+'
    else:
        return 'Fail'

df['Grade'] = df['Total Marks'].apply(grading_system)
```

```python
# i also want to check if the students get less than 90 marks in any subject then they will fail. 
# so i will use the following code from Grade column to check if the student has failed in any subject.
# If the student has failed in any subject, then the student will fail overall.
# plus absent in any exam will also be considered as fail.
# so i will use the following code to check if the student has failed in any subject or absent in any exam.

df = pd.read_csv('students.csv')
def absence_means_fail(row):

    # subjects = df.iloc[row.name, 5:9]  # Get the subject marks for the current row
    subjects =df.columns[5:9]  # Get the subject columns (Python, System Design, DBMS, Networking, Math)
    for sub in subjects:
        if pd.isna(row[sub]) or row['Total Marks'] < 90:  # Check for NaN or marks less than 90
            return 'Fail'

    return 'Pass'

df['absent_means_fail'] = df.apply(absence_means_fail, axis=1)
df.head(45)
```
