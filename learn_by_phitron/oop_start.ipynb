```python
# python properties is a way to define and control access to attributes in a class. it is a decorator
# it is a built-in decorator that allows us to define getter, setter, and deleter methods for an attribute in a class.
# different between getter and setter methods and properties are, 
# getter and setter methods are defined using regular methods, 
# while properties are defined using the @property decorator. 
# Properties provide a more concise and readable way to define getter and setter methods for an attribute.
# we use properties for syntax cleanliness and future-proofing.

class User:  
    def __init__(self, username, email, password):
        self.username = username
        self._email = email # protected
        self.__password = password  #private      

    @property # here we define a property called email, which allows us to access the _email attribute using the email property instead of the get_email() method.
    def email(self): # here we define a getter method for the email property
        return self._email.lower().strip()
    @email.setter  # here we define a setter method for the email property, which allows us to modify the _email attribute using the email property instead of the set_email() method.
    def email(self, new_email):
        if "@" not in new_email:
            raise ValueError("Invalid email address")
        self._email = new_email

# so we do not need to use get_email() and set_email() methods anymore, we can use the email property instead.
# The email property provides a clean and safe way to access and modify the _email attribute.
# usind set_email() and get_email() methods is not recommended, we should use the email property instead. 
# because it is more concise and readable, and it provides a more pythonic way to access and modify the _email attribute.

user1 = User("Akter", " Akter@gmail.com    ", "djfhlaksdjfhlkasjd49084ß30958" )

print(user1.email) # this will call the email property and return the cleaned email address

#now we will update the email using the email property instead of the set_email() method
user1.email = "193@outlook.com" # this will call the email property and update the _email attribute
print(f"Updated email: {user1.email}") # this will call the email property and return the cleaned email address
```

```python
# overall property is efficient and clean way to define getter and setter methods for an attribute in a class.
class QueryExecution:

  def __init__(self, query, attempts):
    self.query = query
    self.attempts = attempts  # Triggers the setter below!

  @property
  def attempts(self):
    return self._attempts

  @attempts.setter
  def attempts(self, value):
    if not 1 <= value <= 5:
      raise ValueError("Attempts must be between 1 and 5!")
    self._attempts = value


# --- Testing it ---
run = QueryExecution("SELECT ...", 4)  # Works fine
print(run.attempts)  # Output: 4

# This will now safely trigger an error and protect your data:
# run.attempts = 10  # Raises ValueError: Attempts must be between 1 and 5!
```

```python
# instance methods are methods that operate on an instance of a class and can access the instance's attributes and other methods.
# easy: instance methods are defined using the def keyword and take self as the first parameter, which refers to the instance of the class.
# differece between instance and static methods is that instance methods can access and modify the instance's attributes, while static methods cannot.
# static methods are defined using the @staticmethod decorator and do not take self as the first parameter. 
# They are used for utility functions that do not need to access or modify the instance's attributes.

class MDXQuery:
    # __init__ runs each time we create a new MDXQuery object.
    def __init__(self, question, mdx_query):
        # self.question belongs to this particular object.
        self.question = question

        # self.mdx_query also belongs to this particular object.
        self.mdx_query = mdx_query

    # INSTANCE METHOD: it receives self, so it can use this object's data.
    def show_details(self):
        return f"Question: {self.question}\nMDX query: {self.mdx_query}"

    # STATIC METHOD: it does not receive self.
    # We give it a query string directly, and it checks that string.
    @staticmethod
    def has_basic_structure(query):
        query = query.strip().upper()

        # This is only a simple text check, not a full MDX validator.
        return query.startswith("SELECT ") and " FROM " in query


# Create one instance (object) of the MDXQuery class.
result_1 = MDXQuery(
    "What are the total internet sales?",
    "SELECT {[Measures].[Internet Total Sales]} ON COLUMNS "
    "FROM [Adventure Works Internet Sales Model]"
)

# Call the INSTANCE method on the object.
# It knows which question and query to display because of self.
print(result_1.show_details())

# Call the STATIC method on the class.
# We must pass the query string because this method has no self.
is_valid = MDXQuery.has_basic_structure(result_1.mdx_query)
print(f"Has basic MDX structure: {is_valid}")
```

```python
# encapsulation is the packing of data and functions into a single component and restricting access to some of the object's components.
# write the code 
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance  #The double underscore prefix makes the balance attribute private.

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}. New balance: {self.__balance}")
        else:
            print("Invalid withdrawal amount.")

    def get_balance(self):
        return self.__balance

private_account = BankAccount("123456789", 1000)
private_account.deposit(500)
private_account.withdraw(200)
print(f"Current balance: {private_account.get_balance()}")

try:
    print(private_account.__balance)  # This will raise an AttributeError because __balance is private and cannot be accessed directly.
except AttributeError as e:
    print(f"Error: {e}")

# how did we use encapsulation in general: it is a fundamental concept in object-oriented programming (OOP) 
# that helps to protect the internal state of an object and control how it is accessed and modified. 
# We have encapsulated the balance attribute by making it private (using double underscores) 
# and provided public methods (deposit, withdraw, and get_balance) to interact with it safely.
# what is happening here is that the balance attribute is private and cannot be accessed directly from outside the class.
# Instead, we provide public methods (deposit, withdraw, and get_balance) to interact with the balance safely.
```

```python
# Abstraction is the concept of hiding the complex implementation details and showing only the essential features of an object.
# write the code with abstraction and explain it
from abc import ABC, abstractmethod # why ABC and abstractmethod are used here is because we want to create an abstract class that cannot be instantiated directly, 
# but can be subclassed.

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod # here easy to understand - abstractmethod decorator is used to define an abstract method that must be implemented by any subclass of the Animal class.
    def make_sound(self):
        pass # why pass is used here is because we do not want to provide an implementation for the make_sound() method in the Animal class.

class Dog(Animal):
    def make_sound(self):
        return f"{self.name} barks: Woof! Woof!"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} meows: Meow! Meow!"

# Creating instances of the classes
dog = Dog("Buddy")
cat = Cat("Whiskers")

# Using the abstract method
print(dog.make_sound())
print(cat.make_sound())

# here , we have an abstract class Animal that defines a blueprint for animals. 
# It has an abstract method make_sound() that must be implemented by any subclass.
# The Dog and Cat classes inherit from Animal and provide their own implementations of the make_sound() method.
# This allows us to create specific animal types while enforcing a common interface for making sounds.

# Abstraction is the concept of hiding the complex implementation details and showing only the essential features of an object.
# write the code with abstraction and explain it
```
