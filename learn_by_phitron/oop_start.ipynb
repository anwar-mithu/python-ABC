```python
class Result: # here Result is class (Parent)
    def __init__(self, question, mdx_query, executed):  # question, mdx_query, and executed are attributes. values down below > 
        self.question = question # Inside the class, self means the particular instance using this method.
        self.mdx_query = mdx_query
        self.executed = executed
        # The plain question on the right is a parameter of __init__ 
        # self.question on the left is the instance attribute.

    def show_status(self): # methods: A method is function defined inside a class. it usually works with that instance's attributes
        print(self.mdx_query)
        if self.executed == True:
            print("Operation Successful")
        else:
            print("Unsucessful")

# result_1 is an instance or object
result_1 = Result(
    "What are the internet sales by year?", # these are the values of the question attribute
    "SELECT ... hf. FROM [Adventure Works]", # these are the values of the mdx_query attribute
    True # these are the values of the executed attribute
)

# result_2 is an instance or object
result_2 = Result(
    "What are the internet sales by month?",
    "SELECT ... khb.  ON ROWS FROM [Adventure Works]",
    False
)

# Class = definition. Instance = one actual item made using that definition.
```

```python
print(result_1.question)
print(result_1.mdx_query)
print(result_1.executed)

# result_1 → the instance
# question → the attribute name
# "What are the sales by year?" → the value stored in that attribute
```

```python
result_2.show_status() # it shows the status of attributes
```

```python
# child class
class Execution(Result): 
    def __init__(self, question, mdx_query, executed, time, token, tools):
        super().__init__(question, mdx_query, executed)
        self.time = time
        self.token = token
        self.tools = tools

result_1_execution = Execution(
    result_1.question,
    result_1.mdx_query,
    result_1.executed,
    120,
    12000,
    2
)

result_2_execution = Execution(
    result_2.question,
    result_2.mdx_query,
    result_2.executed,
    140,
    1000,
    4
)

print(result_1_execution)

print(result_2_execution.time)
```

```python
#polymorphism refers to the way in which different object classes can share the same method name,
#  but those methods can act differently based on which object calls them.
class MDXResult:
    def summary(self):
        return "MDX result from an OLAP cube"


class SQLResult:
    def summary(self):
        return "SQL result from a database"


results = [MDXResult(), SQLResult()]

for result in results:
    print(result.summary())
```

```python
# encapsulation is the packing of data and functions into a single component and restricting access to some of the object's components.
# write the code 
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.__balance = balance  # Private attribute and protected attribute are the same. The double underscore prefix makes the balance attribute private.

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}. New balance: {self.__balance}")
        else:
            print("Invalid withdrawal amount.")

    def get_balance(self):
        return self.__balance

private_account = BankAccount("123456789", 1000)
private_account.deposit(500)
private_account.withdraw(200)
print(f"Current balance: {private_account.get_balance()}")

try:
    print(private_account.__balance)  # This will raise an AttributeError because __balance is private and cannot be accessed directly.
except AttributeError as e:
    print(f"Error: {e}")

# what is happening here is that the balance attribute is private and cannot be accessed directly from outside the class.
# Instead, we provide public methods (deposit, withdraw, and get_balance) to interact with the balance safely.
```

```python
# Abstraction is the concept of hiding the complex implementation details and showing only the essential features of an object.
# write the code with abstraction and explain it



from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod # here easy to understand - abstractmethod decorator is used to define an abstract method that must be implemented by any subclass of the Animal class.
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return f"{self.name} barks: Woof! Woof!"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} meows: Meow! Meow!"

# Creating instances of the classes
dog = Dog("Buddy")
cat = Cat("Whiskers")

# Using the abstract method
print(dog.make_sound())
print(cat.make_sound())

# here , we have an abstract class Animal that defines a blueprint for animals. 
# It has an abstract method make_sound() that must be implemented by any subclass.
# The Dog and Cat classes inherit from Animal and provide their own implementations of the make_sound() method.
# This allows us to create specific animal types while enforcing a common interface for making sounds.
```
