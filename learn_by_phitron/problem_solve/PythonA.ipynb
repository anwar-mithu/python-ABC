```python
_in_space = "4 6 8 12 16 20 24 28"

_in_string = _in_space.split()

_in_int = []

for number in _in_string: 
    _in_int.append(int(number))

print(_in_int)

output_list = []

for i in _in_int: 
    if i % 4 == 0 and i % 8 != 0:
        output_list.append(i)
    else:
        pass

print(output_list)        

```

```python
def compute_ratio(x,y):
    answer =  (x + y) / (x * y)
    return answer

try: 
    print(compute_ratio(10,10))
except Exception as e: 
    print(f"The error cause by: {e}")

finally: 
    print("Execution completed")
```

```python
with open("data.txt", "w") as file:
    file.write("Python is a wonderful language for machine learning.")

content = open("data.txt", "r").read()
content = content.rstrip(".")
print(content)

_sep = content.split()

word = ""
longest_word = ""

for word in _sep:
    if len(word) > len(longest_word):
        longest_word = word

print(longest_word)
```

```python
text = "An apple a day keeps the doctor away also an orrange is good"

text_in_list = text.split()

print(text_in_list)

vowel_list = []

for words in text_in_list:
    words = words.lower()
    if words[0] == 'a' or words[0] == 'e' or words[0] == 'i' or words[0] == 'o' or words[0] == 'u': 
        vowel_list.append(words)

print(vowel_list)
```

```python
input_word = "Fax"

input_word = input_word.lower()

reversed_word = input_word[::-1]

if input_word == reversed_word:
    print("Palindrome")
else:
    print("Not Palindrome")
```
