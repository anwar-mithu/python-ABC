```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

```python

print("Employee Ages Array: ", emp_ages)
print("Employee Income Array: ", emp_income)

print(f"Data Type: {type(emp_ages)}",)
print(f"Data Type: {type(emp_income)}",)

print(f"Number of Dimensions: {emp_ages.ndim}")
print(f"Number of Dimensions: {emp_income.ndim}")

print(f"Shape: {emp_ages.shape}")
print(f"Shape: {emp_income.shape}")

print(f"Total Number of Elements for Ages: {emp_ages.size}")
print(f"Total Number of Elements for Income: {emp_income.size}")


 
```

```python
min_salary = np.min(emp_income)
print(f"Minimum Salary: {min_salary}")

mean_salary = np.mean(emp_income)
print(f"Mean Salary: {mean_salary}")

mean_age = np.mean(emp_ages)
print(f"Mean Age: {mean_age}")
```

```python
extracted_employees = np.where((emp_ages >= 25) & (emp_ages <= 40))
print(f"Employees aged between 25 and 40: {emp_ages[extracted_employees]}")

extracted_salaries = np.where((emp_income >= 50000))
print(f"Employees with salary >= 50000: {emp_income[extracted_salaries]}")

len_employees = len(emp_income[extracted_salaries])
print(f"Number of Employees with salary >= 50000: {len_employees}")
```

```python
print(f"All the ages: {emp_ages}")

average_age = np.mean(emp_ages)
print(f"Average Age: {average_age}")

adjusted_ages = np.where(emp_ages < average_age, emp_ages + 4, emp_ages - 2)
print(f"Adjusted Ages: {adjusted_ages}")

new_average_age = np.mean(adjusted_ages)
print(f"New Average Age: {new_average_age}")
```

```python
developers = data[data['role'].str.contains('Developer')].to_numpy()

print(f"Developer Old salaries: {developers[:, 2]}") 

#increase the salaries of developers by 8%
developer_salaries = developers[:, 2].astype(float) 
adjusted_salaries = developer_salaries * 1.08
print(f"Adjusted salaries: {adjusted_salaries}")
```
