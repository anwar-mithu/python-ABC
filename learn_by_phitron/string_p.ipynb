```python
#write a long text
text = "Hello, this is a long text that is meant to demonstrate how to handle and process large strings in Python. It can be used for various purposes such as testing string manipulation functions, analyzing text data, or simply for educational purposes. The text can contain multiple sentences, paragraphs, and even special characters to simulate real-world scenarios. By working with long texts, we can better understand the performance implications and best practices for string handling in Python."
```

```python
print(text[0:100])  # Print the first 100 characters of the text
print(text[-100:])  # Print the last 100 characters of the text
print(len(text))  # Print the length of the text
```

```python
print("Hello" in text)  # Check if "Hello" is in the text
print(text.replace("Hello", "programming"))  # Replace "Python" with "programming"
```

```python
print(split_text := text.split('.'))  # Split the text into sentences
print(f"Number of sentences: {len(split_text)}")
```

```python
processed_text = text.lower()  # Convert the text to uppercase
print(processed_text)  # Print the processed text
```

```python
capitalized_text = text.capitalize()  # Capitalize the first letter of the text
print(capitalized_text)  # Print the capitalized text
```

```python
index = text.find("this")  # Find the index of the word "this"
print(f"The word 'this' is found at index: {index}")
```

```python
ends_with = text.endswith("Python.")  # Check if the text ends with "Python."
print(f"Does the text end with 'Python.'? {ends_with}")
```

```python
#split the text into words
#what is split() method in python
#answer: The split() method in Python is a built-in string method that divides a string into a list of substrings based on a specified delimiter. By default, 
# it splits the string at whitespace characters (spaces, tabs, newlines), but you can specify any character or substring as the delimiter. 
# The result is a list of the separated substrings.
words = text.split()  # Split the text into words   
print(words)  # Print the list of words
print(f"Number of words: {len(words)}")  # Print the number of words
```

```python
# how splits help in tokenization
# Tokenization is the process of breaking down a text into smaller units called tokens, which can be words, phrases, or symbols.
# The split() method helps in tokenization by allowing us to easily divide a string into individual words or tokens based on a specified delimiter. 
# This is particularly useful in natural language processing (NLP) tasks, where we need to analyze and manipulate text data. 
# By splitting the text into tokens, we can perform various operations such as counting word frequencies, 
# identifying unique words, and preparing the text for further analysis or machine learning models.

words = "How are you ?"
#lets tokenize the above text using split() method
tokens = words.split(" ")  # Tokenize the text into words
print(tokens)  # Print the list of tokens
print(f"Number of tokens: {len(tokens)}")  # Print the number of tokens
```

```python
#convert split to joint again
joins = "-".join(tokens)
print(joins)
```
