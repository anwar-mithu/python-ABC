```python
import numpy as np

arr = np.random.randint(1, 100, size=(10, 5)) # row 10, column/feature 5, sample/row 10 
#feature name - age, height, weight, income, score
#sample data - we will generate random data for 10 samples with 5 features
print(arr)
arr.shape
```

```python
b = arr.reshape(5,10)

print(b) # reshape is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 5*10 = 50, so reshape is possible.
print(b.shape)

c = arr.reshape(25,2) # reshape is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 25*2 = 50, so reshape is possible.
print(c)
```

```python
d = arr.reshape(2,25) # reshape is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 2*25 = 50, so reshape is possible.
print(d)

e = arr.reshape(2,5,5) # reshape is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 2*5*5 = 50, so reshape is possible.
print(e)
```

```python
f = arr.reshape(1, 50) # reshape is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 50*1 = 50, so reshape is possible.
print(f)
```

```python
g = arr.flatten() # flatten is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 50*1 = 50, so flatten is possible.
print(g)
```

```python
ct = np.ravel(g, order = 'C')  
 # ravel is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 50*1 = 50, so ravel is possible.
print(ct)   
```

```python
r = np.ravel(g, order = 'F')   # ravel is only possible if the total number of elements remains the same. Here, 10*5 = 50 and 50*1 = 50, so ravel is possible.
print(r)

# F means to read the array in column-major order (Fortran-style), 
# it means that the first index changes the fastest, and the last index changes the slowest.
# while C means to read the array in row-major order (C-style). The default is C.
# it means that the last index changes the fastest, and the first index changes the slowest.
```

```python
# create a 20 students and 5 subjects marks and find the average marks of each student and each subject in pandas

import pandas as pd
import numpy as np
students = ['Student_' + str(i) for i in range(1, 21)]
subjects = ['Subject_' + str(i) for i in range(1, 6)]
marks = np.random.randint(0, 101, size=(20, 5))

df = pd.DataFrame(marks, columns=subjects, index=students)
print(df)

# send this data to csv file and read it
df.to_csv('students_marks.csv', index=True)

```

```python
data = np.genfromtxt('students_marks.csv', delimiter=',', skip_header=1, usecols=range(1, 6))

print(data)
```

```python
who_got_maximum_marks = np.max(data, axis=0)
print(who_got_maximum_marks)
```

```python
subject_2_marks = data[:, 1]
print(subject_2_marks)

subject_2_min = np.min(subject_2_marks)
print(subject_2_min)

subject_2_max = np.max(subject_2_marks)
print(subject_2_max)

subject_2_average = np.mean(subject_2_marks)
print(subject_2_average)
```

```python
subject_4_marks = data[:, 3]
print(subject_4_marks)

subject_4_min = np.min(subject_4_marks)
print(subject_4_min)

subject_4_max = np.max(subject_4_marks)
print(subject_4_max)

subject_4_average = np.mean(subject_4_marks)
print(subject_4_average)

subject_5_std = np.std(data[:, 4])
print(subject_5_std)

# std 26.65 means the standard deviation of the marks in subject 5 is 26.65, 
# which indicates how much the marks deviate from the average marks in that subject.
# A higher standard deviation means that the marks are more spread out, 
# while a lower standard deviation means that the marks are closer to the average.
```
