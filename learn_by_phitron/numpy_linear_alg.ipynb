```python
# linear algebra - matrix multiplication, dot product, inverse, determinant, eigenvalues, eigenvectors, etc.

import numpy as np

mat_A_2d = np.array([[1, 2], [3, 4]])
mat_B_2d = np.array([[5, 6], [7, 8]])

# multiplication
multiplication = np.matmul(mat_A_2d, mat_B_2d)
print("Matrix Multiplication:\n", multiplication)
```

```python
# eigenvectors
mat_C_2d = np.array([[1, 2], [2, 1]])
print("Matrix C:\n", mat_C_2d)
eigenvalues, eigenvectors = np.linalg.eig(mat_C_2d)
print("Eigenvalues of Matrix C:", eigenvalues)  
```

```python
# dot product
dot_product = np.dot(mat_A_2d, mat_B_2d)
print("Dot Product:\n", dot_product)
```

```python
# inverse
mat_A_2d = np.array([[1, 2], [3, 4]])
print("Matrix A:\n", mat_A_2d)
mat_A_inv = np.linalg.inv(mat_A_2d)
print("Inverse of Matrix A:\n", mat_A_inv)
```

```python
#determinant 
mat_B_2d = np.array([[5, 6], [7, 8]])
print("Matrix B:\n", mat_B_2d)
det_B = np.linalg.det(mat_B_2d) 
print("Determinant of Matrix B:", det_B)     
```

```python
# eigenvalues

mat_C_2d = np.array([[1, 2], [2, 1]])
print("Matrix C:\n", mat_C_2d)
eigenvalues, eigenvectors = np.linalg.eig(mat_C_2d)
print("Eigenvalues of Matrix C:", eigenvalues)
print("Eigenvectors of Matrix C:\n", eigenvectors)
```
