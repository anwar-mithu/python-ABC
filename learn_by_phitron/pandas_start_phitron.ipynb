```python
# data clean can be done b pandas
# exploratory data engineering
# feature engineering
# we will learn this ML course

# create a file containing data in csv format - 100 students - with fullname, enrollment number, birthdate, age, DSA, System Design, DBMS, OS, Python marks 
# with delimiter comma
# how to write delimiter comma in csv file -

# do not run this code again, it will overwrite the file and data will be lost
with open('students.csv', 'w') as f:
    f.write('fullname,enrollment_number,birthdate,age,DSA,System_Design,DBMS,OS,Python\n')
    for i in range(1, 101):
        fullname = f'Student {i}'
        enrollment_number = f'ENR{i:03d}'
        birthdate = f'2000-01-{i%30 + 1:02d}'
        age = 2024 - 2000
        DSA = i % 100
        System_Design = (i * 2) % 100
        DBMS = (i * 3) % 100
        OS = (i * 4) % 100
        Python = (i * 5) % 100
        f.write(f'{fullname},{enrollment_number},{birthdate},{age},{DSA},{System_Design},{DBMS},{OS},{Python}\n')
```
