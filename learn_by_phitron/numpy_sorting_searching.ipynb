```python
#searching
searching = np.array([[[34,67,12,34],
                       [123,3445,67,734],
                       [56,7426,78,465]], 
                      [[13,637,1232,324],
                       [123,435,67,74],
                       [56,346,728,43265]]])

#index= np.where(searching == 34)

#print(index)


index_equal= np.where(searching == 12)

print(index_equal)
#  (array([0]), array([0]), array([2])) - element - 0, row - 0, column  - 2(0,1,2)

index_less = np.where(searching < 13)
print(index_less)

# (array([0, 1]), array([0, 0]), array([2, 0])) - we have two value here. understand!!!!!!!!!!!!

```

```python
max_index = np.argmax(searching)

print(f"Index of maximum value: {max_index}") # it will return the index of the maximum value in the flattened array.

max_value = np.max(searching)

print(f"Value of maximum: {max_value}") # it will return the maximum value in the array.

min_index = np.argmin(searching)
print(f"Index of minimum value: {min_index}") # it will return the index of the minimum value in the flattened array.

min_value = np.min(searching)
print(f"Value of minimum: {min_value}") # it will return the minimum value in the array. if we have multiple minimum value it will return the first one.   
```

```python
#nd array counting
our_counting_array = np.array([[[34,67,12,34],    
                      [123,3445,67,734],
                      [56,7426,78,465]], 
                     [[13,637,1232,324],
                      [123,435,67,74],
                      [56,346,728,43265]]])

count_34 = np.count_nonzero(our_counting_array == 34)
print(f"Count of 34: {count_34}") # it will return the count of the number of occurrences of the specified value (34 in this case) in the array.

counting_zero = np.count_nonzero(our_counting_array == 0)
print(f"Count of 0: {counting_zero}") # it will return the count of the number of occurrences of the specified value (0 in this case) in the array.

counting_less_than_100 = np.count_nonzero(our_counting_array < 100)
print(f"Count of values less than 100: {counting_less_than_100}") # it will return the count of the number of occurrences of the values less than 100 in the array.

counting_67 = np.count_nonzero(our_counting_array == 67)
print(f"Count of 67: {counting_67}") # it will return the count of the number of occurrences of the specified value (67 in this case) in the array.
```

```python
unique_values = np.unique(our_counting_array, return_counts=True)
print(f"Unique values: {unique_values[0]}")
print(f"Counts, the number of occurrences of each unique value: {unique_values[1]}") 
# {unique_values[1]} - it means the number of occurrences of each unique value in the array. 
# The first element of the tuple returned by np.unique is an array of the unique values, and the second element is an array of the corresponding counts.
```
