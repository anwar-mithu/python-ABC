```python
dict = {"name": "Alice", "age": 30, "city": "New York"}
dict["age"] = 33  # Update the age to 33
dict["country"] = "USA"  # Add a new key-value pair for country
dict.pop("city")  # Remove the key-value pair for city

print(dict.get("name"))  # Access the value for the key "name"
print(dict)
```

```python
#list inside dictionary

list_dict = {
    "fruits": ["apple", "banana", "cherry"],
    "vegetables": ["carrot", "broccoli", "spinach"],
    "grains": ["rice", "wheat", "corn"], 
    "dairy": ["milk"]
}

list_dict["other"] = ["bags"]  # Add a new fruit to the list

print(list_dict["dairy"])  # Access the list of fruits

keys = list_dict.keys()
values = list_dict.values()
items = list_dict.items()

print(keys)
print(values)
print(items)
```

```python
for key, value in list_dict.items():
    print(f"{key}: {value}")
```

```python
square = {x:x **2 for x in range(10)} # sq = { key : value }
print(square)
```

```python
lo_cation = ["bielefeld", "frankfurt", "Köln"]

co_ordinates = [(10,45,78),(23,67,43),(1,35,88)]

c_ity = {lo_ca : co_or for  lo_ca, co_or in zip(lo_cation, co_ordinates)}

print(c_ity["frankfurt"])

print(c_ity)

set_city = set(c_ity.keys())
list_city = list(c_ity.keys())
print(set_city)
print(list_city)
```

```python
#write a 100 words paragraph with iterable words. a lot of words used in iterable form.

paragraph = "my name is john, his name is mike, her name is sarah, their names are tom and jerry, our names are alice and bob, your names are charlie and dave, its name is rex, these names are lily and rose, those names are jack and jill, every name is unique, some names are common, many names are rare, few names are popular, all names have meanings, each name has a story, any name can be chosen, no name is too long, some names are short, many names are beautiful, few names are simple, all names can be remembered, each name can be cherished, any name can be loved, no name should be forgotten."

split_para = paragraph.split()
print(split_para)

count = {}

for word in split_para: 
    count[word] = count.get(word,0) + 1
print(count)
```
