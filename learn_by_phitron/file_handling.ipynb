```python
#write a new file
with open("new_file.txt", "w") as file:
    file.write("This is a new file created using python\n")
    file.write("This is the second line of the new file\n")
    file.write("This is the third line of the new file\n")


#append some new lines to a file
with open("new_file.txt", "a") as file:
    file.write("This is the fourth line of the new file\n")
    file.write("This is the fifth line of the new file\n")

with open("new_file.txt", "r") as file:
    for line in file: 
        print(line)

```
