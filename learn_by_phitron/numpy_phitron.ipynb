```python
# numpy is a powerful library for numerical computing in Python. 
# It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays efficiently.
list_demo_different_types = [1, 2.5, 'three', True, None]  # A list containing different data types
i = 0
for item in list_demo_different_types:
    print(f"Item {i}: {item}, Type: {type(item)}")  # Print each item and its type
    i += 1
    try:
        x = item + 1
        print(x)
    except TypeError as e:
        print(f"Cannot add 10 to {item} of type {type(item)}: {e}")  # Handle TypeError for unsupported operations

# the pronlem of list is that it can contain elements of different data types, 
# which can lead to unexpected behavior when performing operations on the list. For example,
# trying to add a number to a string or a boolean will raise a TypeError, as shown in the code above. 
# This is why it's important to be aware of the data types of the elements in a list and handle them appropriately to avoid errors.
#### this is why numpy is preferred for numerical computations, 
#### as it enforces a uniform data type for its arrays, allowing for more efficient and predictable operations.
```

```python
import numpy as np 

x = np.array([2,46,1,35,778,21])  # Create a NumPy array with integers

print(type(x))
addition_result = x + 10  # Add 10 to each element of the array
print("Original array:", x)  # Print the original array
print("Array after adding 10:", addition_result)  # Print the result after addition
```

```python
# 1. Array creation and manipulation
import numpy as np
# Create a 1D array

arr_1d = np.array(["apple", "banana", "cherry"])  # Create a 1D NumPy array with strings
print("1D array:", arr_1d)  # Print the 1D array

# Create a 2D array
arr_2d = np.array([["apple", "banana"], ["cherry", "date"]])  # Create a 2D NumPy array with strings
print("2D array:", arr_2d)  # Print the 2D array
print("Shape of 2D array:", arr_2d.shape)  # Print the shape of the 2D array

# Create a 3D array
arr_3d = np.array([[["apple", "banana"], ["cherry", "date"]], [["fig", "grape"], ["kiwi", "lemon"]]])  

print("3D array:", arr_3d)  # Print the 3D array
print("Shape of 3D array:", arr_3d.shape)  # Print the shape of the 3D array
# how it is 3d array? because it has 3 dimensions, the first dimension has 2 elements, each of which is a 2D array with 2 rows and 2 columns.
# what is rows and columns?
# rows are horizontal lines of elements, while columns are vertical lines of elements. 
# In the 3D array above, each 2D array has 2 rows and 2 columns, making it a 3D array overall.

# 4d array
arr_4d = np.array([[[["apple", "banana"], ["cherry", "date"]], [["fig", "grape"], ["kiwi", "lemon"]]], [[["mango", "nectarine"], ["orange", "papaya"]], [["quince", "raspberry"], ["strawberry", "tangerine"]]]])
print("4D array:", arr_4d)  # Print the 4D array
print("Shape of 4D array:", arr_4d.shape)  # Print the shape of the 4D array 
# here the 4D array has 2 elements in the first dimension, 
# each of which is a 3D array with 2 elements in the second dimension, 
# each of which is a 2D array with 2 rows and 2 columns.
```
